# Parcial 2
Simule el problema de una esfera de icopor que caen en el aire, la densidad del icopor es de 20 kg/m^3.
* Muestre en una misma gráfica la posición en función del tiempo  para tres casos r=1mm, r=1cm y r=1dm, debe mostrar explicitamente el tiempo que tarda en caer un metro de distancia en cada caso
* Muestre en una misma gráfica la velocidad en función del tiempo  para tres casos r=1mm, r=1cm y r=1dm, debe mostrar explicitamente la velocidad límite en cada caso

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
g, p, pf, r, Cd = 9.8, 20, 1.2, 0.001, 0.47
a = 3*Cd*pf/(8*r*p)
x0, v0 = 0, 0
R0 = [x0, v0]
def ecuacion(R0, t):
  x, v = R0
  F = [v, (1-(pf/p))*g-a*v**2]
  return F
t = np.linspace(0, 4, 200)
xsol = odeint(ecuacion, R0, t)
i = 0
L1 = []
V1 = xsol[:, 1]
T1 = []
while xsol[i, 0]<1:
  L1.append(xsol[i, 0])
  T1.append(t[i])
  i = i+1
```

```python
g, p, pf, r, Cd = 9.8, 20, 1.2, 0.01, 0.47
a = 3*Cd*pf/(8*r*p)
x0, v0 = 0, 0
R0 = [x0, v0]
def ecuacion(R0, t):
  x, v = R0
  F = [v, (1-(pf/p))*g-a*v**2]
  return F
t = np.linspace(0, 4, 200)
xsol = odeint(ecuacion, R0, t)
i = 0
L2 = []
V2 = xsol[:, 1]
T2 = []
while xsol[i, 0]<1:
  L2.append(xsol[i, 0])
  T2.append(t[i])
  i = i+1
```

```python
g, p, pf, r, Cd = 9.8, 20, 1.2, 0.1, 0.47
a = 3*Cd*pf/(8*r*p)
x0, v0 = 0, 0
R0 = [x0, v0]
def ecuacion(R0, t):
  x, v = R0
  F = [v, (1-(pf/p))*g-a*v**2]
  return F
t = np.linspace(0, 4, 200)
xsol = odeint(ecuacion, R0, t)
i = 0
L3 = []
V3 =  xsol[:, 1]
T3 = []
while xsol[i, 0]<1:
  L3.append(xsol[i, 0])
  T3.append(t[i])
  i = i+1
```

```python
plt.plot(T1, L1, label='Distancia recorrida para r=1mm')
plt.plot(T2, L2, label='Distancia recorrida para r=1cm')
plt.plot(T3, L3, label='Distancia recorrida para r=1dm')
plt.legend()
plt.grid()
plt.xlabel("tiempo (s)")
plt.ylabel("Distancia (m)")
plt.show()
print("El tiempo que tarda en recorrer 1m es: ")
print(round(T1[-1],2), "segundos para r=1mm")
print(round(T2[-1],2), "segundos para r=1cm")
print(round(T3[-1],2), "segundos para r=1dm")
```

```python
plt.plot(t, V1, label='Velocidad para r=1mm')
plt.plot(t, V2, label='Velocidad para r=1cm')
plt.plot(t, V3, label='Velocidad para r=1dm')
plt.legend()
plt.grid()
plt.xlabel("tiempo (s)")
plt.ylabel("Velocidad en (m/s)")
plt.show()
print("La velocidad límite es: ")
print(round(V1[-1],2), "m/s para r=1mm")
print(round(V2[-1],2), "m/s para r=1cm")
print(round(V3[-1],2), "m/s para r=1dm")
```
